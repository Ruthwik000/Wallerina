```python
import requests
import json
import re
import time


SEARCH_URL = "https://gamma-api.polymarket.com/public-search"
PRICE_URL = "https://clob.polymarket.com/prices-history"


# ============================================================
# ASSET -> POLYMARKET SEARCH TERMS
# ============================================================

SEARCH_TERMS = {
    "ETH": ["ethereum", "ether"],
    "DOGE": ["dogecoin"],
    "OP": ["optimism"],
    "POL": ["polygon"],
    "CAKE": ["pancakeswap"],
    "MOG": ["mog"],
}


# ============================================================
# SEARCH POLYMARKET
# ============================================================

def search_polymarket(query):

    response = requests.get(
        SEARCH_URL,
        params={
            "q": query,
            "limit_per_type": 20,
            "events_status": "active"
        },
        timeout=30
    )

    print(
        f"Search '{query}' -> HTTP {response.status_code}"
    )

    response.raise_for_status()

    return response.json()


# ============================================================
# PARSE JSON-STRING FIELDS
# ============================================================

def parse_json(value):

    if isinstance(value, str):

        try:
            return json.loads(value)
        except:
            return value

    return value


# ============================================================
# GET YES TOKEN + PRICE
# ============================================================

def extract_yes_data(market):

    outcomes = parse_json(
        market.get("outcomes", [])
    )

    prices = parse_json(
        market.get("outcomePrices", [])
    )

    token_ids = parse_json(
        market.get("clobTokenIds", [])
    )

    if not isinstance(outcomes, list):
        return None

    for i, outcome in enumerate(outcomes):

        if str(outcome).lower() != "yes":
            continue

        if i >= len(token_ids):
            return None

        probability = None

        if i < len(prices):

            try:
                probability = float(
                    prices[i]
                )
            except:
                pass

        return {
            "yes_token": token_ids[i],
            "probability": probability
        }

    return None


# ============================================================
# 5-MINUTE PRICE CHANGE
# ============================================================

def get_5m_change(token_id):

    now = int(time.time())

    response = requests.get(
        PRICE_URL,
        params={
            "market": token_id,
            "startTs": now - 300,
            "endTs": now,
            "fidelity": 1
        },
        timeout=15
    )

    if response.status_code != 200:
        return None

    history = response.json().get(
        "history",
        []
    )

    if len(history) < 2:
        return None

    try:

        first = float(history[0]["p"])
        last = float(history[-1]["p"])

        if first == 0:
            return None

        return (last - first) / first

    except:
        return None


# ============================================================
# FIND RELEVANT MARKETS
# ============================================================

def get_asset_markets(asset):

    results = []
    seen = set()

    for query in SEARCH_TERMS[asset]:

        data = search_polymarket(query)

        for event in data.get("events", []):

            event_title = event.get(
                "title",
                ""
            )

            event_text = (
                event_title + " " +
                event.get("description", "")
            ).lower()

            for market in event.get(
                "markets",
                []
            ):

                if market.get("closed"):
                    continue

                question = market.get(
                    "question",
                    ""
                )

                text = (
                    question + " " +
                    event_text
                ).lower()

                # ------------------------------------------------
                # Avoid false positives
                # ------------------------------------------------

                if asset == "ETH":

                    relevant = (
                        "ethereum" in text or
                        re.search(r"\bether\b", text)
                    )

                elif asset == "DOGE":

                    relevant = (
                        "dogecoin" in text
                    )

                elif asset == "OP":

                    relevant = (
                        "optimism" in text
                    )

                elif asset == "POL":

                    relevant = (
                        "polygon" in text
                    )

                elif asset == "CAKE":

                    relevant = (
                        "pancakeswap" in text
                    )

                elif asset == "MOG":

                    relevant = bool(
                        re.search(
                            r"\bmog\b",
                            text
                        )
                    )

                else:
                    relevant = False

                if not relevant:
                    continue

                market_id = market.get(
                    "id"
                )

                if market_id in seen:
                    continue

                seen.add(market_id)

                yes_data = extract_yes_data(
                    market
                )

                if yes_data is None:
                    continue

                try:
                    volume = float(
                        market.get("volume") or 0
                    )
                except:
                    volume = 0

                try:
                    liquidity = float(
                        market.get("liquidity") or 0
                    )
                except:
                    liquidity = 0

                results.append({
                    "question": question,
                    "event": event_title,
                    "probability": yes_data["probability"],
                    "yes_token": yes_data["yes_token"],
                    "volume": volume,
                    "liquidity": liquidity
                })

    # Highest-liquidity markets first
    results.sort(
        key=lambda x: (
            x["liquidity"],
            x["volume"]
        ),
        reverse=True
    )

    return results


# ============================================================
# RUN FOR WALLET ASSETS
# ============================================================

polymarket_data = {}

print("=" * 80)
print("POLYMARKET DATA")
print("=" * 80)


for asset in sorted(wallet_assets):

    print()
    print(asset)

    markets = get_asset_markets(
        asset
    )

    if not markets:

        print("  No relevant market")

        polymarket_data[asset] = []

        continue


    # Keep top 10
    markets = markets[:10]

    polymarket_data[asset] = []

    for market in markets:

        change_5m = get_5m_change(
            market["yes_token"]
        )

        market["change_5m"] = change_5m

        polymarket_data[asset].append(
            market
        )

        print()
        print(
            "  Market:",
            market["question"]
        )

        print(
            f"  Probability: "
            f"{market['probability'] * 100:.2f}%"
        )

        if change_5m is not None:

            print(
                f"  5m change: "
                f"{change_5m * 100:+.2f}%"
            )

        print(
            f"  Liquidity: "
            f"${market['liquidity']:,.2f}"
        )

        print(
            f"  Volume: "
            f"${market['volume']:,.2f}"
        )


# ============================================================
# CLEAN MONTE CARLO STRUCTURE
# ============================================================

print()
print("=" * 80)
print("MONTE CARLO INPUT")
print("=" * 80)

for asset, markets in polymarket_data.items():

    print(f"\n{asset}")

    for m in markets:

        print({
            "probability": m["probability"],
            "change_5m": m["change_5m"],
            "liquidity": m["liquidity"],
            "volume": m["volume"],
            "question": m["question"]
        })
```
